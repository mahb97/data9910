can also use python's floor division operator directly //

```python
45782 // 227
```

int() truncates toward zero (chops off the decimal part, no matter the sign), while // does floor division (rounds down toward negative infinity). For positive numbers, they behave identically but when dividing negative numbers, they can disagree. E.g. -7 / 2 = -3.5; int(-3.5) gives -3, but -7 // 2 gives -4. 

This is exactly the logic behind computing something like "how many complete 227-word chunks fit into a 45,782-word novel" for a sliding-window stylometric analysis. Don't want a fractional chunk, but want to know how many whole windows you get, and floor division is precisely the tool for that.

5. **Create a list for each of *ints*, *floats* and *strings*. Add every item in *items* to the correct list** 

```python
items = [5, 1.5, 'hello', 7, 2.0, 3, 'world']

items_int = []
items_float = []
items_str = []

for ele in items:
    if type(ele) == int:
        items_int.append(ele)
    elif type(ele) == float:
        items_float.append(ele)
    elif type(ele) == str:
        items_str.append(ele)

print(items_int)
print(items_float)
print(items_str)
```

this is exactly the logic of a token classifier sorting a corpus into bins. When iterating over a list of tokens and routing each one into a punctuation, word, or number bucket based on its category, before running category-specific processing on each bucket (you wouldn't lowercase a number or strip punctuation from a word the same way). The elif chain here is functioning like a decision tree / simple rule-based tagger: check the most specific test first, fall through if it doesn't match, and each item lands in exactly one bin.

8. **Write a function called isOdd. The function should take a single parameter and return True if it's an odd number, false otherwise**

```python
def isOdd(num):
    return num % 2 != 0

print(isOdd(7))   # True
print(isOdd(4))   # False
print(isOdd(0))   # False
print(isOdd(-3))  # True
```

```python
def isOdd(num):
    if num % 2 != 0:
        return True
    else:
        return False

print(isOdd(7))   # True
print(isOdd(4))   # False
print(isOdd(0))   # False
print(isOdd(-3))  # True
```

this is structurally identical to writing a binary classifier function. Like isFunctionWord(token) returning True/False based on whether a token belongs to a closed class (determiners, prepositions, etc.). The building block you'd chain together with filter() or a list comprehension to pull matching items out of a larger corpus, the same way you might isolate odd numbers from a list, or function words from a token stream.

9. **Update the isOdd function. If the user passes any value that isn't an int return False**

```python
def isOdd(num):
    if type(num) != int:
        return False
    return num % 2 != 0

print(isOdd(7))      # True
print(isOdd(4))      # False
print(isOdd('7'))    # False  (it's a string, not an int)
print(isOdd(3.0))    # False  (it's a float, not an int)
print(isOdd([1,2]))  # False  (it's a list, not an int)
```

10. ***Write a function called limit100. The function should take a single numeric parameter, the function should return the parameter provided it's less than 100. If the number is greater than 100, the function returns 100. Make sure you test it with some values***

```python
def limit100(num):
    if num > 100:
        return 100
    return num

print(limit100(50))    # 50   (below 100, unchanged)
print(limit100(150))   # 100  (above 100, capped)
print(limit100(100))   # 100  (exactly 100, unchanged, happens to equal the cap anyway)
print(limit100(-20))   # -20  (negative numbers pass through fine, no lower cap specified)
print(limit100(99.9))  # 99.9 (works fine on floats too, since the comparison isn't type-restricted)
```

same logic as value clipping/clamping in normalization pipelines. e.g. when plotting word frequencies or similarity scores on a fixed-scale visualization and want to cap any extreme outlier (a hapax-adjacent word with a wildly disproportionate frequency spike, say) at some ceiling so it doesn't distort your axis, you'd apply exactly this kind of "cap at threshold, otherwise pass through" logic. It's also the core idea behind min()-based clipping. This same function could be written more tersely as return min(num, 100), which does the identical comparison-and-cap in one call.

11. **Write a lambda function to produce a list which halves the values of items**

```python
items = [20, 18, 16, 14, 12, 10, 8, 6, 4, 2, 0]

halved = list(map(lambda x: x / 2, items))
print(halved)
```

lambda x: x / 2 defines a tiny unnamed function that takes one input (x) and returns x / 2. It's equivalent to writing:

  def half(x):
      return x / 2

map() + lambda is exactly the shape of applying a per-token transformation across a whole text. e.g. list(map(lambda tok: tok.lower(), tokens)) to lowercase every token in a corpus, or list(map(lambda w: len(w), words)) to get a list of word lengths for a lexical-complexity measure. The pattern is always the same: one small transformation rule, applied uniformly across every element of a sequence, without writing an explicit for loop.

18. **Fix the ordering of this list using insert() and remove() to put everything in its correct order**

```python
notes = ['Do', 'Re', 'Do', 'Fa', 'Mi', 'Ti', 'La', 'So']

# Step 1: pull both 'Do's out and re-add them at the correct positions (start and end) once everything else is in order.
notes.remove('Do')
notes.remove('Do')
# notes is now: ['Re', 'Fa', 'Mi', 'Ti', 'La', 'So']

# Step 2: fix Fa/Mi as Mi should come before Fa
notes.remove('Fa')
notes.insert(2, 'Fa')
# notes is now: ['Re', 'Mi', 'Fa', 'Ti', 'La', 'So']

# Step 3: fix So as it should come right after Fa, not at the end
notes.remove('So')
notes.insert(3, 'So')
# notes is now: ['Re', 'Mi', 'Fa', 'So', 'Ti', 'La']

# Step 4: fix Ti/La as La should come before Ti
notes.remove('Ti')
notes.append('Ti')
# notes is now: ['Re', 'Mi', 'Fa', 'So', 'La', 'Ti']

# Step 5: put 'Do' back at the start and the end
notes.insert(0, 'Do')
notes.append('Do')

print(notes)
```

structurally similar to fixing word-order scrambling in a sentence-reordering task (the kind used in psycholinguistic studies of syntactic processing, or in evaluating whether an LLM can unscramble a shuffled sentence back into canonical word order) where you're not changing what the tokens are, only where they sit relative to one another, item by item, until the sequence satisfies the target grammar (here, "diatonic scale order" instead of "syntax"). The duplicate-'Do' wrinkle is a nice parallel to handling repeated tokens in a sentence (e.g., "the dog chased the dog") where algorithms that reorder or align sequences by value alone need extra care whenever a token type isn't unique, since position, not just identity, is what actually needs tracking.

## Sets 

22. **Create a set, A, containing the elements 'a', 'b' and 'c'**

```python
A = {'a', 'b', 'c'}
print(A)
```

{} on its own creates an empty dictionary, not an empty set. Python reserves the bare {} for dicts, since dicts are far more commonly used. If you want an empty set, you have to use the set() constructor instead:

empty_set = set()

Sets are the natural data structure for representing a vocabulary (as opposed to a token list). If tokens is your ordered list of every word occurrence in a text (with repeats, since "the" shows up hundreds of times), set(tokens) collapses that down to just the distinct word types, which is exactly the type/token distinction that's foundational in corpus linguistics. Set operations then map directly onto comparative vocabulary questions: the intersection of two authors' vocabularies gives you shared lexis, the difference gives you what's unique to one author but not the other (this is genuinely how a lot of basic stylometric vocabulary-overlap analysis gets implemented).

24. **Check the length of the list and the set from 25. above. Why are they different lengths?**

```python
l = ['c', 'c', 'd', 'e', 'f']
B = set(l)

print(len(l))
print(len(B))
```

l is a list with 5 elements including the repeated 'c' but lists don't care about duplicates, they just store items in the order you gave them. B, on the other hand, is a set, and sets automatically enforce uniqueness: when you built B = set(l), the second 'c' was silently dropped because a set can only contain each distinct value once. So len(l) counts all 5 occurrences, while len(B) counts only the 4 distinct values.
